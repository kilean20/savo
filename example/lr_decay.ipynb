```python
import numpy as np
import matplotlib.pyplot as plt

k = 10.0  # Steepness (default)
y0 = 0.8 # Midpoint (default)
y = np.linspace(0, 1, 128)
reduction = 1 / (1 + np.exp(k * (y - y0)))


# --- Plotting ---
plt.figure(figsize=(8, 6))
plt.plot(y, reduction, label=f'lr')
plt.plot(y, reduction**2, label=f'lrES')
plt.xlabel("y")
plt.ylabel("Function Value")
plt.grid(True)
plt.axvline(x=y0, color='r', linestyle='--', label=f'Midpoint y₀ = {y0}') # Show midpoint
plt.legend()
plt.show()
```
